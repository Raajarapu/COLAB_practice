```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load Titanic dataset
df = sns.load_dataset("titanic")

# Display the first few rows
print(df.head())

# Check for missing values
print(df.isnull().sum())
```

```python
df['age']=df['age'].fillna(df['age'].mean())
```

```python
df['deck'] = df['deck'].fillna(df['deck'].mode()[0])
```

```python
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df.isnull().sum()
```

# Line Plot

```python
plt.figure(figsize=(10, 5))
plt.plot(df['age'], df['fare'], marker='o', linestyle='-', color='blue', alpha=0.7)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Line Plot: Age vs. Fare")
plt.grid(True)
plt.show()
```

# Scatter Plot

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(x=df['age'], y=df['fare'], hue=df['sex'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Scatter Plot: Age vs. Fare")
plt.show()
```

# Bar Chart

```python
plt.figure(figsize=(10, 5))
sns.barplot(x="class", y="fare", data=df, estimator=np.mean, palette="viridis")
plt.title("Bar Chart: Average Fare by Class")
plt.show()
```

# Histogram

```python
plt.figure(figsize=(10, 5))
plt.hist(df['age'].dropna(), bins=20, color='purple', alpha=0.7, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Histogram: Age Distribution")
plt.show()
```

# Box Plot

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x="class", y="fare", data=df, palette="Set2")
plt.title("Box Plot: Fare Distribution by Class")
plt.show()
```

# Pie Chart

```python
plt.figure(figsize=(6, 6))
df['sex'].value_counts().plot.pie(autopct='%1.1f%%', colors=['lightblue', 'pink'])
plt.title("Pie Chart: Male vs Female Distribution")
plt.ylabel("")  # Remove y-axis label
plt.show()
```

# Stacked Area Plot

```python
plt.figure(figsize=(10, 5))
plt.stackplot(df['age'].dropna(), df['fare'].dropna(), alpha=0.5, labels=['Fare'])
plt.legend(loc="upper left")
plt.title("Stacked Area Plot: Age vs Fare")
plt.show()
```

# 3D Scatter Plot

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(df['age'], df['fare'], df['pclass'], c='r', marker='o')
ax.set_xlabel("Age")
ax.set_ylabel("Fare")
ax.set_zlabel("Passenger Class")
ax.set_title("3D Scatter Plot: Age vs Fare vs Class")
plt.show()
```

# Clustered Heatmap

**Dendrogram**

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(df[['age', 'fare']].dropna(), method='ward')
plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title("Dendrogram: Hierarchical Clustering")
plt.show()
```

# Chord Diagram (Network Relationships)

```python
import networkx as nx

G = nx.Graph()
edges = df.groupby(["embark_town", "class"]).size().reset_index()
edges.columns = ["embark_town", "class", "weight"]

for _, row in edges.iterrows():
    G.add_edge(row["embark_town"], row["class"], weight=row["weight"])

plt.figure(figsize=(8, 6))
nx.draw(G, with_labels=True, node_color="lightblue", edge_color="gray", node_size=2000, font_size=10)
plt.title("Network Graph: Embark Town & Class Connections")
plt.show()
```

# Lollipop Chart

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df["class"], y=df["fare"], color="red", s=100)
plt.vlines(x=df["class"], ymin=0, ymax=df["fare"], color="gray")
plt.title("Lollipop Chart: Fare by Class")
plt.show()
```

# Slope Chart

```python
df_slope = df.groupby("class")[["fare"]].mean().reset_index()

plt.figure(figsize=(6, 5))
plt.plot(["Class 1", "Class 2", "Class 3"], df_slope["fare"], marker="o", color="blue")
plt.title("Slope Chart: Fare Change Across Classes")
plt.show()
```

# Bullet Chart

```python
plt.figure(figsize=(8, 3))
plt.barh(["Target Fare"], [50], color="gray", alpha=0.3)
plt.barh(["Target Fare"], [df["fare"].mean()], color="blue")
plt.title("Bullet Chart: Average Fare vs Target")
plt.show()
```

# nnotated Line Chart

```python
plt.figure(figsize=(8, 5))
plt.plot(df["fare"].sort_values().reset_index(drop=True), color="purple")
plt.annotate("Highest Fare", xy=(df["fare"].idxmax(), df["fare"].max()), xytext=(100, 200), arrowprops=dict(facecolor='red'))
plt.title("Annotated Line Chart: Fare Trend")
plt.show()
```
