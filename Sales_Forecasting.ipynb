## Data Cleaning and Preparation

```python
# Convert 'Order_Date' and 'Ship_Date' to datetime objects
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d/%m/%Y', errors='coerce')
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], format='%d/%m/%Y', errors='coerce')

# Check for any NaT (Not a Time) values introduced by coercion
print(f"NaT values in Order_Date after conversion: {df['Order_Date'].isna().sum()}")
print(f"NaT values in Ship_Date after conversion: {df['Ship_Date'].isna().sum()}")

# Handle missing Postal_Code values
# Given only 11 missing values out of 9800, we can drop these rows.
df.dropna(subset=['Postal_Code'], inplace=True)
# Convert Postal_Code to integer type after dropping NaNs
df['Postal_Code'] = df['Postal_Code'].astype(int)

print("\nDataFrame Info after initial cleaning:")
df.info()
```

## Feature Engineering: Extracting Temporal Features

```python
# Extract year, month, day, and day of week from Order_Date
df['Order_Year'] = df['Order_Date'].dt.year
df['Order_Month'] = df['Order_Date'].dt.month
df['Order_Day'] = df['Order_Date'].dt.day
df['Order_DayOfWeek'] = df['Order_Date'].dt.dayofweek # Monday=0, Sunday=6

# Calculate Order_Processing_Days
df['Order_Processing_Days'] = (df['Ship_Date'] - df['Order_Date']).dt.days

# Display the first few rows with new features
display(df[['Order_Date', 'Ship_Date', 'Order_Year', 'Order_Month', 'Order_Day', 'Order_DayOfWeek', 'Order_Processing_Days']].head())
```

## Model Building: Data Preparation

```python
# Define target variable and features
# We will try to predict 'Sales'

# Drop irrelevant columns for modeling
# 'Row_ID', 'Order_ID', 'Customer_ID', 'Customer_Name', 'Product_ID', 'Product_Name' are identifiers or descriptive text.
# 'Order_Date' and 'Ship_Date' have already been used to derive temporal features.
columns_to_drop = [
    'Row_ID', 'Order_ID', 'Customer_ID', 'Customer_Name', 'Product_ID',
    'Product_Name', 'Order_Date', 'Ship_Date'
]
df_model = df.drop(columns=columns_to_drop, errors='ignore')

# Separate target variable (y) and features (X)
X = df_model.drop('Sales', axis=1)
y = df_model['Sales']

print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}")
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split

# Identify categorical and numerical columns for preprocessing
categorical_cols = list(X.select_dtypes(include='object').columns) # Ensure list of strings
numerical_cols = list(X.select_dtypes(include=['int64', 'float64']).columns) # Ensure list of strings

# Create a column transformer for one-hot encoding categorical features
# Explicitly handle both categorical and numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', 'passthrough', numerical_cols) # Explicitly pass numerical columns
    ],
    remainder='drop' # Explicitly drop any other columns not mentioned
)

# Apply preprocessing
X_processed = preprocessor.fit_transform(X)

# Convert to dense array if it's a sparse matrix
if hasattr(X_processed, "toarray"):
    X_processed = X_processed.toarray()

# Get feature names after one-hot encoding directly from the ColumnTransformer
all_feature_names = preprocessor.get_feature_names_out()

# Convert X_processed to DataFrame for better readability
X_processed_df = pd.DataFrame(X_processed, columns=all_feature_names, index=X.index)

print(f"Shape of X after one-hot encoding: {X_processed_df.shape}")
display(X_processed_df.head())
```

## Splitting Data into Training and Testing Sets

```python
# Split the processed data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_processed_df, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

## Model Training

```python
from sklearn.ensemble import RandomForestRegressor

# Initialize the Random Forest Regressor model
# Using default parameters for initial training, can be tuned later
model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)

# Train the model
print("Training the Random Forest Regressor...")
model.fit(X_train, y_train)
print("Model training complete.")
```

## Model Evaluation

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R2): {r2:.2f}")

# Plot actual vs. predicted sales
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.6)
plt.title('Actual vs. Predicted Sales')
plt.xlabel('Actual Sales')
plt.ylabel('Predicted Sales')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', linestyle='--') # Diagonal line
plt.show()
```
